### 4.4 Model Selection

**4.4 Answer:**  
The models we tried for this dataset were Linear Regression, Random Forest, and Gradient Boosting models. These models were selected due to the relationships between the variables of interest. The data of interest was mainly numeric, which works well with regression analysis. Categorical analysis was ignored due to few categorical variables being present within the original dataset. Dr. Snyder also confirmed that these models were the best to evaluate our dataset in class. 

The generalization of the model was evaluated in two ways. First, the model was split into a test set and a training set using TrainTestSplit. We separated 20% of the dataset for the testing set and the remaining 80% of the dataset was used for testing. We also used 5-fold cross validation within our models. This allowed the model to be evaluated in more granular chunks, thus allowing each of the data points to be evaluated. This provided a stable basis for selecting optimal hyperparameters. 

For hyperparameter tuning, three methods were used. The first method was sweep_parameters. With this method, we tuned each parameter starting with wide searches with low granularity and progressively increasing repeats while decreasing search area. The results were collected in a dictionary and fed into the model. The second method used for hyperparameter tuning was RandomSearch. This method randomly tests sets of parameters from a range. It then tests the predictions within the models. The final method used for tuning our models was Optuna. This method is an automated search that uses a study and trials to test various hyperparameters and outputs the best results. Notably, grid search was not utilized due to the high computational costs and long computation times associated with the method. 

The final model was chosen by evaluating the key metrics output from the tuned models. We used both mean average error and standard deviation to evaluate our models. We chose a model that minimizes the mean average error and maintains a small standard deviation. This means that the model’s output is both accurate and consistent.

The lower the MAE is, the more accurate the model will be to the actual value. This was chosen for this model because we preprocessed this data to remove outliers. That means the majority of the data can be understood through MAE, as Root Mean Square Error (RMSE) will penalize outliers heavily. We would use RMSE if there were large outliers still present in the dataset in order to penalize the outliers. MAE also allows for better readability and interpretability of the data for shareholders/clients of this project. Standard deviation was measured so that the consistency of the model can be evaluated. A smaller standard deviation shows that data deviates less from the mean, which would imply that the model is more consistent. By combining the MAE and Standard Deviation, and selecting a model that has minimized both values, we can choose a model that is both accurate and consistent in predicting data. 

The key metrics described above were used to evaluate the models and the hyperparameter tuning. It was found that the Gradient Boosting model has the best baseline MAE and Standard Deviation. After feature engineering, the Gradient Boosting model was still outperforming the other two models. While both of the key metrics here were not the lowest, they were both nearly the lowest of all of the models.
