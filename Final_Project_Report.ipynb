### 4.3 Feature Engineering

Describe your feature engineering.

Include:
- transformations or new features
- why you created them
- which were useful or not useful
- what worked and what did not work

**4.3 Answer:**  
In this project, I applied feature engineering to try to improve model performance and better capture patterns in the housing data. Based on earlier analysis, I created three new features: log_calculatedfinishedsquarefeet, sqft_per_room, and bath_bed_ratio.

The first transformation was applying a log transformation to the square footage variable. Housing size data tends to be skewed, so taking the log helps make the distribution more balanced. This is especially useful for models like Linear Regression, since they perform better when relationships are more linear. Because of that, this feature helped the model better understand how square footage relates to property value.

The second feature, sqft_per_room, was created to represent how space is distributed within a home. Instead of just looking at total square footage or number of rooms separately, this gives a better sense of how much space each room actually has. The idea was that two homes with the same size could feel very different depending on how that space is divided.

The third feature, bath_bed_ratio, was meant to capture the relationship between bathrooms and bedrooms. This was included to reflect potential differences in home quality, since properties with more bathrooms per bedroom are often considered more desirable.

After creating these features, I scaled the data using StandardScaler so that all variables were on a similar scale. Then I re-ran the models using repeated cross validation to see if these new features improved performance.

The results were mixed. Linear Regression showed a small improvement, with MAE decreasing from about 153276 to 152977. This suggests that the log transformation was helpful, since it made the data easier for the model to work with. On the other hand, Random Forest and Hist Gradient Boosting did not improve and actually performed slightly worse.

This was not too surprising. Ensemble models like Random Forest and Gradient Boosting are already good at capturing complex relationships and interactions between variables, so adding engineered features does not always give them much new information. Overall, the new features did not hurt performance, but they also did not lead to major improvements for the stronger models.

In the end, the most useful feature was the log transformation of square footage, especially for Linear Regression. The other features were reasonable ideas, but they did not add much beyond what the original variables were already capturing. This shows that feature engineering can help, but its impact depends on the type of model being used.


### 4.4 Model Selection

Describe how you selected your model.

Include:
- models you tried and why
- how you evaluated generalization
- hyperparameter tuning
- how you chose the final model

**4.4 Answer:**  
The models we tried for this dataset were Linear Regression, Random Forest, and Gradient Boosting models. These models were selected due to the relationships between the variables of interest. The data of interest was mainly numeric, which works well with regression analysis. Categorical analysis was ignored due to few categorical variables being present within the original dataset. Dr. Snyder also confirmed that these models were the best to evaluate our dataset in class.

The generalization of the model was evaluated in two ways. First, the model was split into a test set and a training set using TrainTestSplit. We separated 20% of the dataset for the testing set and the remaining 80% of the dataset was used for testing. We also used 5-fold cross validation within our models. This allowed the model to be evaluated in more granular chunks, thus allowing each of the data points to be evaluated. This provided a stable basis for selecting optimal hyperparameters.

For hyperparameter tuning, three methods were used. The first method was sweep_parameters. With this method, we tuned each parameter starting with wide searches with low granularity and progressively increasing repeats while decreasing search area. The results were collected in a dictionary and fed into the model. The second method used for hyperparameter tuning was RandomSearch. This method randomly tests sets of parameters from a range. It then tests the predictions within the models. The final method used for tuning our models was Optuna. This method is an automated search that uses a study and trials to test various hyperparameters and outputs the best results. Notably, grid search was not utilized due to the high computational costs and long computation times associated with the method.

The final model was chosen by evaluating the key metrics output from the tuned models. We used both mean average error and standard deviation to evaluate our models. We chose a model that minimizes the mean average error and maintains a small standard deviation. This means that the model’s output is both accurate and consistent.
